````python
#import all necessary labraries
import requests
from bs4 import BeautifulSoup
import re
import pandas as pd
import numpy as np
````

## **Scrapping Movies Data**

````python
# Specifying the URL from which movies related data will be fetched
url='https://www.justwatch.com/in/movies?release_year_from=2000'

# Sending an HTTP GET request to the URL
page=requests.get(url)
# Parsing the HTML content using BeautifulSoup with the 'html.parser'
soup=BeautifulSoup(page.text,'html.parser')
# Printing the prettified HTML content
print(soup.prettify())
````

## **Fetching Movie URL's**

````python
# Write Your Code here
# Finding all <a> tags with class 'title-list-grid__item--link' which contain movie URLs
movie_links = soup.find_all('a', class_='title-list-grid__item--link')

# Extracting URLs from the href attribute of <a> tags and concatenating the base URL
base_url = 'https://www.justwatch.com'
movie_urls = [base_url + link['href'] for link in movie_links]

# Printing the extracted movie URLs
for url in movie_urls:
    print(url)
````

## **Scrapping Movie Title**

````python
# Write Your Code here

movie_names = []

# Iterate over each URL to scrape movie names
for url in movie_urls:
    # Send an HTTP GET request to the URL
    page = requests.get(url)

    # Parse the HTML content using BeautifulSoup
    soup = BeautifulSoup(page.text, 'html.parser')

    # Find the element containing the movie name
    movie_name_element = soup.find('h1')

    # Extract the text of the movie name
    if movie_name_element:
        movie_names.append(movie_name_element.text.strip())
    else:
        movie_names.append('Movie name not found')

# Print the movie names
for name in movie_names:
    print(name)
````

## **Scrapping release Year**

````python
# Write Your Code here
movie_release_years = {}

# Iterate over each URL to scrape movie names and release years
for url in movie_urls:
    # Send an HTTP GET request to the URL
    page = requests.get(url)

    # Parse the HTML content using BeautifulSoup
    soup = BeautifulSoup(page.text, 'html.parser')

    # Find the element containing the movie name
    movie_name_element = soup.find('h1')
    movie_name = movie_name_element.text.strip() if movie_name_element else 'Movie name not found'

    # Find the element containing the release year
    release_year_element = soup.find('span', class_='text-muted')
    release_year = release_year_element.text.strip().strip('()') if release_year_element else 'Release year not found'

    # Add movie name and release year to the dictionary
    movie_release_years[movie_name] = release_year

# Print movie names and their release years
for name, year in movie_release_years.items():
    print(f"{name} : {year}")
````

## **Scrapping Genres**

````python
generes = {}

for url in movie_urls:
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    movie_name_element = soup.find('h1')
    movie_name = movie_name_element.text.strip()

    genere = soup.select_one('.detail-infos__subheading:contains("Genres") + .detail-infos__value').text.split(',')
    genere=[i.strip()for i in genere]
    if genere:
        generes[movie_name] = genere
    else:
        generes[movie_name] = "No rating found"

for name, rating in generes.items():
    print(f"{name} :{genere}")
````

## **Scrapping IMBD Rating**

> Indented block



````python
ratings = {}

for url in movie_urls:
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    movie_name_element = soup.find('h1')
    movie_name = movie_name_element.text.strip()

    rating_element = soup.select_one('.jw-scoring-listing__rating span')

    if rating_element:
        rating_text = rating_element.text.strip().split()[0]
        ratings[movie_name] = rating_text
    else:
        ratings[movie_name] = "No rating found"

#print(ratings)
for name, rating in ratings.items():
    print(f"{name} : {rating}")
````

## **Scrapping Runtime/Duration**

````python
# Write Your Code here
runtimes = {}

for url in movie_urls:
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    movie_name = soup.select_one('h1').text.strip()

    runtime = soup.select_one('.detail-infos__subheading:contains("Runtime") + .detail-infos__value').text

    if runtime:
       runtimes[movie_name] = runtime
    else:
      runtimes[movie_name] = "No runtime found"

for name, runtime in runtimes.items():
     print(f"{name} :{runtime}")
````

## **Scrapping Age Rating**

````python
# Write Your Code here
age_ratings = {}

for url in movie_urls:
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    movie_name = soup.select_one('h1').text.strip()

    age_rating = soup.select_one('.detail-infos__subheading:contains("Age rating") + .detail-infos__value')

    if age_rating:
       age_ratings[movie_name] = age_rating.text.strip()
    else:
       age_ratings[movie_name] = "No age rating found"

for name, age_rating in age_ratings.items():
     print(f"{name} : {age_rating}")
````

## **Fetching Production Countries Details**

````python
# Write Your Code here
prod_country = {}

for url in movie_urls:
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    movie_name = soup.select_one('h1').text.strip()

    country = soup.select_one('.detail-infos__subheading:contains("Production country") + .detail-infos__value').text.strip().split(',')

    if prod_country:
       prod_country[movie_name] = country
    else:
       prod_country[movie_name] = "No runtime found"

for name, country in prod_country.items():
     print(f"{name} :{country}")
````

## **Fetching Streaming Service Details**

````python
# Write Your Code here
streaming_mdetail = {}

for url in movie_urls:
    response = requests.get(url)
    soup = BeautifulSoup(response.text, 'html.parser')

    movie_name = soup.select_one('h1').text.strip()

    streaming_m1 = soup.select_one('h2.heading + p').text.strip()

    streaming_m2=streaming_m1.split(" ")
    streaming=[]
    for i in streaming_m2:
        if i == 'Amazon':
             streaming.append("Amazon Prime Video")
        if i == 'Jio':
             streaming.append("Jio Cinema")
        if i == 'Crunchyroll':
             streaming.append("Crunchyroll")
        if i == 'Hotstar' or i == 'Hotstar.':
            streaming.append("Hotstar")
        if i == 'Netflix' or i == 'Netflix.':
            streaming.append("Netflix")
        if i == 'Voot' or i == 'Voot.':
            streaming.append("Voot")
        if i == 'MX':
            streaming.append("MX Player")
        if i == 'Sony':
            streaming.append("Sony Liv")
        if i == 'Alt':
            streaming.append("Alt Balaji")
        if i == 'Apple':
            streaming.append("Apple TV")
        if i == 'VI':
            streaming.append("VI movies")
        if i == 'Zee5' or i=='Zee5.':
            streaming.append("Zee5")
        if i == 'constantly':
            streaming.append("N/A")

        streaming_mdetail[movie_name] = list(set(streaming))

for movie_name, streaming in streaming_mdetail.items():
     print(f"{movie_name} : {streaming}")
````

````python
# Write Your Code here

import pandas as pd
import numpy as np

list_of_dict= [movie_release_years,generes,ratings,runtimes,age_ratings,prod_country,streaming_mdetail]
df_movies = pd.DataFrame(list_of_dict, index = ['Release year', 'Genre','IMDB rating','Duration','Age rating','Country of Production','Streaming details']).transpose()
df_movies.head(50)

````
